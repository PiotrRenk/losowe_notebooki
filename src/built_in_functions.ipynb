## @property

Python doesnt use getter-setter methods, instead it uses @property -> which is an equivalent to getter method and the extensions of it


https://docs.python.org/2/library/functions.html#property

```python
class C(object):
    def __init__(self):
        self._x = None

    @property
    def x(self):
        """I'm the 'x' property."""
        return self._x

    @x.setter
    def x(self, value):
        self._x = value #note i can f.e. add 10 to the value and return 20 if value = 10

    @x.deleter
    def x(self):
        print("Deleting x... ") # (note i do not need to delete it, i can f.e. add 10 here)
        del self._x
```

```python
obj = C()
print(obj.x) # Output: None
obj.x = 10
print(obj.x) # Output: 10
del obj.x # calls the delete  fuction
try:
    print(obj.x) # Output: None
except AttributeError as e:
    print(e)
```

## all(iterable)

Return True if all elements of the iterable are true (or if the iterable is empty)

```python
print(all([True, True, True])) # Output: True
print(all([True, False, True])) # Output: False
res = all(lambda: x > 0 for x in [1,2,3])
print(res) # Output: True
```

Equivalent to:

```python
def all(iterable):
    for element in iterable:
        if not element:
            return False
    return True
```
